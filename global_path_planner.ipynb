### Начинаем движение из узла 0 и ищем путь до узла 6. 
### Для этого выполняем алгоритм Беллмана-Форда и находим все кратчайшие пути от узла 0 до всех других узлов. Здесь мы получаем массив предков в котором порядковый номер элемента соответствует номеру узла, а содержащееся значение говорит о предшественнике.

```python
predecessor_array = g.BellmanFord(7)
print(predecessor_array)
```

### Затем восстанавливем путь по графу от узла 0 до 6.

```python
def get_path(src, dist, predecessor_array):
    
    path = []
    
    reached = False
    
    path.append(dist)
    prev = dist
    while not reached:
        path.append(predecessor_array[prev])
        prev = predecessor_array[prev]
        
        
    return path
    
```
